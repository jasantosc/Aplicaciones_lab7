```python
Director = df.STARS.iloc[3].split("|")[0].replace("Director", "").strip()
Stars = df.STARS.iloc[3].split("|")[1].replace("Stars", "").replace("\n", "").strip()
```

```python
def extraer_director(stars_str):
    if "Director" in stars_str:
        clean_str = stars_str.split("|")[0].replace("Director", "").replace("\n", "").strip()
    else:
        return np.nan
```

```python
df["DIRECTOR"] = df["STARS"].apply(extraer_director)
```

```python
def extraer_stars(stars_str):
    if "STAR" in stars_str or "STARS" in stars_str:
        clean_str = stars_str.split("|")[1].replace("STAR", "").replace("STARS", "").replace("\n", "").strip()
    else:
        return np.nan
```

```python
df["STAR"] = df["STARS"].apply(extraer_stars)
```
