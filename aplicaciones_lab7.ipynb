```python
Director = df.STARS.iloc[3].split("|")[0].replace("Director", "").strip()
Stars = df.STARS.iloc[3].split("|")[1].replace("Stars", "").replace("\n", "").strip()
```

```python
def extraer_director(stars_str):
    if "Director" in stars_str:
        clean_str = stars_str.split("|")[0].replace("Director", "").replace("\n", "").strip()
    else:
        return np.nan
```

```python
df["DIRECTOR"] = df["STARS"].apply(extraer_director)
```

```python
def extraer_reparto(stars_str):
    if "STAR" in stars_str or "STARS" in stars_str:
        clean_str = stars_str.split("|")[1].replace("STAR", "").replace("STARS", "").replace("\n", "").strip()
    else:
        return np.nan
```

```python
df["STAR"] = df["STARS"].apply(extraer_reparto)
```

4. Cantidad de registros vacíos en RUNTIME

```python
df["MOVIES"] = df["MOVIES"].str.strip()

print("Nulos en RunTime: ", df["RunTime"].isnull().sum())

rt = df["RunTime"]
a = rt.dropna()
b = rt.fillna(rt.mean())
c = rt.fillna(rt.median())

comparacion = pd.DataFrame({
    "num": [a.count(), b.count(), c.count()],
    "media": [a.mean(), b.mean(), c.mean()],
    "mediana": [a.median(), b.median(), c.median()],
    "std": [a.std(), b.std(), c.std()]
}, index=["Borrar", "Media", "Mediana"]).round(2)

print(comparacion)
print(comparacion)
 
print("Asimetría (skew):", round(rt.skew(), 2))
 

print("RATING promedio (todo):", round(df["RATING"].mean(), 2))
print("RATING promedio (sin nulos RunTime):", round(df.loc[rt.notna(), "RATING"].mean(), 2))
print("% de nulos con RATING = 0:", round(df.loc[rt.isna(), "RATING"].eq(0).mean() * 100, 1))
 
df["RunTime"] = df["RunTime"].fillna(rt.median()) 
```

La mejor opción es la mediana, porque RUNTIME está muy sesgada a la derecha (asimetría de 3,3). Esos valores extremos inflan la media, mientras que la mediana es robusta.
